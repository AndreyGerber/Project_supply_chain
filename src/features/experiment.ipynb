```python
from langdetect import detect, DetectorFactory
DetectorFactory.seed = 0  # für reproduzierbare Ergebnisse

def detect_lang(text):
    try:
        return detect(text)
    except:
        return "unknown"

#df["language"] = df["review_text"].astype(str).apply(detect_lang)
df["language"] = [detect_lang(t) for t in df["review_text"].astype(str)]
print(df["language"].value_counts())

df_not_en=df[df["language"]!="en"]
df_en=df[df["language"]=="en"]
```
